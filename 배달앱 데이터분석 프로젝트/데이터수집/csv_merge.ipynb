```python
import pandas as pd
import glob
```

```python
files = glob.glob(r"../데이터/전처리데이터/*preprocessed*.csv")
files
```

```python
dfs = [pd.read_csv(f, encoding="utf-8-sig") for f in files]
df_all = pd.concat(dfs, ignore_index=True)
len(df_all)
```

```python
df_all.columns
```

```python
df_all.head()
```

```python
df_all['filtered_keywords'] = df_all['filtered_keywords'].apply(lambda x : x.translate(str.maketrans("","","[],'")).split())
df_all.head()
```

```python
df_all['filtered_keywords'][0][0]
```

```python
df_all.to_csv("../데이터/merged_reviews_playstore.csv", index=False, encoding="utf-8-sig")
```
